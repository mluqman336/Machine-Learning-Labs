```python
import pandas as pd

# Load the four CSV files
pricing = pd.read_csv('pricing.csv')
weather = pd.read_csv('weather.csv')
carbon = pd.read_csv('carbon_intensity.csv')
building = pd.read_csv('Building_2.csv')

# Create datetime range: 8760 hours for a year (non-leap)
datetime_range = pd.date_range(start='2024-01-01', periods=8760, freq='H')

# Extract one column from each file (adjust these if needed)
pricing_data = pricing[['electricity_pricing']]
weather_data = weather[['outdoor_dry_bulb_temperature']]
carbon_data = carbon[['carbon_intensity']]
building_data = building[['non_shiftable_load']]

# Add the datetime column to each
pricing_data['Datetime'] = datetime_range
weather_data['Datetime'] = datetime_range
carbon_data['Datetime'] = datetime_range
building_data['Datetime'] = datetime_range

# Rename the data columns to a common name for merging
pricing_data = pricing_data.rename(columns={'electricity_pricing': 'AEP_MW'})
weather_data = weather_data.rename(columns={'outdoor_dry_bulb_temperature': 'AEP_MW'})
carbon_data = carbon_data.rename(columns={'carbon_intensity': 'AEP_MW'})
building_data = building_data.rename(columns={'non_shiftable_load': 'AEP_MW'})

# Stack all data vertically
merged_df = pd.concat([pricing_data, weather_data, carbon_data, building_data], ignore_index=True)

# Reorder columns
merged_df = merged_df[['Datetime', 'AEP_MW']]

# Save to CSV
merged_df.to_csv('merged_two_column_data.csv', index=False)

print("✅ Merged dataset created successfully. Shape:", merged_df.shape)
```
